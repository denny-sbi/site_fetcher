```python
import re
from datetime import datetime
def parse_timestamp(ts_str):
    """Parse timestamps like '5/13/2025 8: 15 AM' into a valid datetime object."""
    try:
        # Remove any extra spaces around colons
        ts_str = re.sub(r"\s*:\s*", ":", ts_str)
        ts_str = re.sub(r"\s+", " ", ts_str).strip()
        return datetime.strptime(ts_str, "%m/%d/%Y %I:%M %p")
    except Exception as e:
        print(f"❌ Error parsing timestamp: {ts_str} - {e}")
        return None
```
